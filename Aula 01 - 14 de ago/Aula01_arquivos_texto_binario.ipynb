## 3. Comparando modo texto e modo binário

```python
with open("arquivo.txt", "r", encoding="utf-8") as arquivo:
    texto = arquivo.read()

print("MODO TEXTO")
print(texto)
print("Tipo:", type(texto))

print("-" * 50)

with open("arquivo.txt", "rb") as arquivo:
    dados = arquivo.read()

print("MODO BINÁRIO")
print(dados)
print("Tipo:", type(dados))
```

## 4. Gerenciamento de arquivos com `with`

O `with` garante que o arquivo seja fechado automaticamente ao sair do bloco.

```python
with open("arquivo.txt", "r", encoding="utf-8") as arquivo:
    conteudo = arquivo.read()
    print(conteudo)

print("Arquivo fechado?", arquivo.closed)
```

### Sem `with`

```python
arquivo = open("arquivo.txt", "r", encoding="utf-8")
conteudo = arquivo.read()

print(conteudo)

arquivo.close()

print("Arquivo fechado?", arquivo.closed)
```

## 8. Caracteres e bytes

```python
texto = "á"

print("Caractere:", texto)
print("Quantidade de caracteres:", len(texto))

dados = texto.encode("utf-8")

print("Representação em bytes:", dados)
print("Quantidade de bytes:", len(dados))
```

### Comparando diferentes caracteres em UTF-8

```python
caracteres = ["A", "á", "ç", "漢", "😊"]

for caractere in caracteres:
    dados = caractere.encode("utf-8")
    print(
        caractere,
        "->",
        dados,
        "->",
        len(dados),
        "byte(s)"
    )
```

## 9. `encode()` e `decode()`

```python
texto = "Olá Ceará"

print("Original:")
print(texto)
print(type(texto))

dados = texto.encode("utf-8")

print("\nApós encode:")
print(dados)
print(type(dados))

texto_novamente = dados.decode("utf-8")

print("\nApós decode:")
print(texto_novamente)
print(type(texto_novamente))
```

## 11. Conversão de Latin-1 para UTF-8

```python
with open("original.txt", "r", encoding="latin-1") as origem:
    texto = origem.read()

with open("convertido.txt", "w", encoding="utf-8") as destino:
    destino.write(texto)

print("Conversão concluída.")

with open("convertido.txt", "r", encoding="utf-8") as arquivo:
    print(arquivo.read())
```

## 13. Posição do cursor com `tell()`

```python
with open("arquivo.txt", "rb") as arquivo:
    print("Posição inicial:", arquivo.tell())

    dados = arquivo.read(5)
    print("Bytes lidos:", dados)

    print("Posição depois da leitura:", arquivo.tell())
```

## 19. Tratamento de erros

```python
try:
    with open("arquivo_inexistente.txt", "r", encoding="utf-8") as arquivo:
        conteudo = arquivo.read()

    print(conteudo)

except FileNotFoundError:
    print("Erro: arquivo não encontrado.")
```

### Tratando diferentes erros

```python
nome = "arquivo.txt"

try:
    with open(nome, "r", encoding="utf-8") as arquivo:
        print(arquivo.read())

except FileNotFoundError:
    print("Arquivo não encontrado.")

except PermissionError:
    print("Sem permissão para acessar o arquivo.")

except UnicodeDecodeError:
    print("A codificação utilizada não corresponde ao arquivo.")

except OSError as erro:
    print("Erro ao acessar o arquivo:", erro)
```

## 24. Detecção manual de BOM

```python
BOM_UTF8 = b"\xef\xbb\xbf"

with open("arquivo_bom.txt", "rb") as arquivo:
    primeiros_bytes = arquivo.read(3)

    if primeiros_bytes == BOM_UTF8:
        print("BOM UTF-8 detectada.")
        conteudo = arquivo.read().decode("utf-8")
    else:
        print("Arquivo sem BOM.")
        arquivo.seek(0)
        conteudo = arquivo.read().decode("utf-8")

print("Conteúdo:", conteudo)
```

## 31. Persistindo um número inteiro em binário com `struct`

```python
import struct

numero = 100

with open("numero.bin", "wb") as arquivo:
    arquivo.write(struct.pack("i", numero))

print("Número salvo em formato binário.")
```

### Recuperando o número

```python
import struct

with open("numero.bin", "rb") as arquivo:
    dados = arquivo.read(4)

numero = struct.unpack("i", dados)[0]

print("Número recuperado:", numero)
```

# Atividade 1 — Lista de Tarefas

O programa deve permitir:

1. adicionar uma tarefa;
2. visualizar tarefas.

Requisitos:

- `with open(...)`;
- modo `a` para adicionar;
- modo `r` para visualizar;
- UTF-8;
- tratar `FileNotFoundError`.

```python
ARQUIVO = "tarefas.txt"

print("=== LISTA DE TAREFAS ===")
print("1 - Adicionar tarefa")
print("2 - Visualizar tarefas")

opcao = input("Escolha uma opção: ")

if opcao == "1":

    descricao = input("Descrição da tarefa: ")

    with open(ARQUIVO, "a", encoding="utf-8") as arquivo:
        arquivo.write(descricao + "\n")

    print("Tarefa adicionada com sucesso.")

elif opcao == "2":

    try:
        with open(ARQUIVO, "r", encoding="utf-8") as arquivo:

            print("\n=== TAREFAS ===")

            for numero, linha in enumerate(arquivo, start=1):
                print(f"{numero}. {linha.strip()}")

    except FileNotFoundError:
        print("Ainda não existem tarefas cadastradas.")

else:
    print("Opção inválida.")
```

# Atividade 1 — Versão com menu contínuo

```python
ARQUIVO = "tarefas.txt"

while True:

    print("\n=== LISTA DE TAREFAS ===")
    print("1 - Adicionar")
    print("2 - Visualizar")
    print("0 - Sair")

    opcao = input("Opção: ")

    if opcao == "1":

        descricao = input("Descrição: ")

        with open(ARQUIVO, "a", encoding="utf-8") as arquivo:
            arquivo.write(descricao + "\n")

        print("Tarefa cadastrada.")

    elif opcao == "2":

        try:
            with open(ARQUIVO, "r", encoding="utf-8") as arquivo:

                print("\n--- TAREFAS ---")

                for numero, tarefa in enumerate(arquivo, start=1):
                    print(f"{numero}. {tarefa.strip()}")

        except FileNotFoundError:
            print("Nenhuma tarefa cadastrada.")

    elif opcao == "0":

        print("Programa encerrado.")
        break

    else:
        print("Opção inválida.")
```

# Atividade 2 — Gerenciador de Tarefas

Funcionalidades:

- adicionar tarefa;
- informar prioridade;
- visualizar;
- buscar;
- remover;
- sair.

```python
ARQUIVO = "tarefas_completas.txt"


def adicionar_tarefa():
    descricao = input("Descrição da tarefa: ").strip()

    prioridade = input(
        "Prioridade (baixa, média ou alta): "
    ).strip().lower()

    if prioridade not in ["baixa", "média", "media", "alta"]:
        print("Prioridade inválida.")
        return

    if prioridade == "media":
        prioridade = "média"

    with open(ARQUIVO, "a", encoding="utf-8") as arquivo:
        arquivo.write(f"{descricao} | {prioridade}\n")

    print("Tarefa adicionada com sucesso.")


def visualizar_tarefas():
    try:
        with open(ARQUIVO, "r", encoding="utf-8") as arquivo:
            tarefas = arquivo.readlines()

        if not tarefas:
            print("Nenhuma tarefa cadastrada.")
            return

        print("\n=== TAREFAS ===")

        for indice, tarefa in enumerate(tarefas, start=1):
            descricao, prioridade = tarefa.strip().split(" | ")

            print(
                f"{indice}. {descricao} "
                f"[Prioridade: {prioridade}]"
            )

    except FileNotFoundError:
        print("Nenhuma tarefa cadastrada.")


def buscar_tarefas():
    palavra = input("Palavra-chave: ").strip().lower()

    try:
        with open(ARQUIVO, "r", encoding="utf-8") as arquivo:
            tarefas = arquivo.readlines()

        encontrou = False

        print("\n=== RESULTADO DA BUSCA ===")

        for indice, tarefa in enumerate(tarefas, start=1):
            descricao, prioridade = tarefa.strip().split(" | ")

            if palavra in descricao.lower():
                print(
                    f"{indice}. {descricao} "
                    f"[Prioridade: {prioridade}]"
                )
                encontrou = True

        if not encontrou:
            print("Nenhuma tarefa encontrada.")

    except FileNotFoundError:
        print("Nenhuma tarefa cadastrada.")


def remover_tarefa():
    try:
        with open(ARQUIVO, "r", encoding="utf-8") as arquivo:
            tarefas = arquivo.readlines()

        if not tarefas:
            print("Nenhuma tarefa cadastrada.")
            return

        print("\n=== TAREFAS ===")

        for indice, tarefa in enumerate(tarefas, start=1):
            print(f"{indice}. {tarefa.strip()}")

        numero = int(
            input("Número da tarefa que deseja remover: ")
        )

        indice = numero - 1

        if indice < 0 or indice >= len(tarefas):
            print("Número de tarefa inválido.")
            return

        tarefa_removida = tarefas.pop(indice)

        with open(ARQUIVO, "w", encoding="utf-8") as arquivo:
            arquivo.writelines(tarefas)

        print(
            "Tarefa removida:",
            tarefa_removida.strip()
        )

    except FileNotFoundError:
        print("Nenhuma tarefa cadastrada.")

    except ValueError:
        print("Digite um número válido.")


def mostrar_menu():
    print("\n==============================")
    print("     GERENCIADOR DE TAREFAS")
    print("==============================")
    print("1 - Adicionar tarefa")
    print("2 - Visualizar tarefas")
    print("3 - Buscar tarefa")
    print("4 - Remover tarefa")
    print("0 - Sair")


while True:

    mostrar_menu()

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        adicionar_tarefa()

    elif opcao == "2":
        visualizar_tarefas()

    elif opcao == "3":
        buscar_tarefas()

    elif opcao == "4":
        remover_tarefa()

    elif opcao == "0":
        print("Programa encerrado.")
        break

    else:
        print("Opção inválida.")
```

# Fluxo conceitual:

**Persistência → Arquivo Texto → Arquivo Estruturado → Arquivo Binário → Banco de Dados**

Pontos principais:

- texto é interpretado utilizando uma codificação;
- arquivos binários são sequências de bytes;
- `with` automatiza o fechamento dos recursos;
- `r`, `w`, `a`, `x`, `b` e `+` alteram a forma de acesso;
- UTF-8 é uma codificação de caracteres;
- `encode()` transforma `str` em `bytes`;
- `decode()` transforma `bytes` em `str`;
- arquivos simples funcionam bem para pequenas persistências;
- operações mais complexas começam a evidenciar a necessidade de mecanismos de persistência mais estruturados.
